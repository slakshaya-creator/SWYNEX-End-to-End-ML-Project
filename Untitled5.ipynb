```python
# =====================================================================================
# Drug Review Sentiment and Side-Effect Mining
# README, data loading, preprocessing, EDA, feature engineering, model training + tuning,
#         cross-validation, evaluation, robustness check, model saving, prediction, user input, results.
# Research and education only. Not medical advice.
# =====================================================================================
import io, re, os, json, html, time, zipfile, warnings
from pathlib import Path
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
import requests, joblib
from IPython.display import Markdown, display
from sklearn.base import BaseEstimator, TransformerMixin, clone
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import ComplementNB
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import label_binarize
from sklearn.model_selection import (train_test_split, GridSearchCV, StratifiedKFold, cross_val_score,
                                     GroupShuffleSplit)
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score, precision_score, recall_score,
                             roc_auc_score, roc_curve, matthews_corrcoef, confusion_matrix, classification_report)
warnings.filterwarnings("ignore"); sns.set_theme(style="whitegrid")

# ------------------------------- SETTINGS -------------------------------
OWN_FILE     = None      # e.g. "/content/drugsComTrain_raw.tsv" (skips the download)
MAX_ROWS     = 30000     # reviews used (full data ~215K; more rows = slower)
DO_TUNING    = True      # small grid search per model
TUNE_SAMPLE  = 10000     # rows used for the grid search
CV_FOLDS     = 3         # stratified cross-validation folds (train set only)
N_BOOT       = 300       # bootstrap resamples for the confidence interval
TEST_SIZE    = 0.20
RANDOM_STATE = 42
INTERACTIVE  = True      # True = ask the user to type reviews at the end
OUT_DIR      = Path("task4_outputs"); OUT_DIR.mkdir(exist_ok=True)
LABELS = ["negative", "neutral", "positive"]
UCI_URLS = ["https://archive.ics.uci.edu/ml/machine-learning-databases/00462/drugsCom_raw.zip",
            "https://archive.ics.uci.edu/static/public/462/drug+review+dataset+drugs+com.zip"]

# ------------------------------- README -------------------------------
README = """
# Drug Review Sentiment and Side-Effect Mining

> **Disclaimer:** research and education only. Not medical advice and not a pharmacovigilance system.

## 1. Overview
Built on the UCI Drug Review dataset (Drugs.com patient reviews). The project predicts whether a review is
**negative, neutral or positive** and extracts the **side effects** mentioned, taking negation into account
(so "no nausea" is not reported as a side effect).

## 2. Installation and how to run
**Google Colab (recommended):** paste this whole cell into one cell and run it. Libraries used (scikit-learn, pandas,
numpy, matplotlib, seaborn, requests, joblib) are preinstalled. The dataset downloads automatically; if that fails,
a file-upload button appears for `drugsComTrain_raw.tsv`. A run on 30,000 reviews takes roughly 5 to 10 minutes.
**Own computer:** Python 3.9+, `pip install numpy pandas scikit-learn matplotlib seaborn requests joblib jupyter`.
**Settings** are at the top of the cell (`MAX_ROWS`, `DO_TUNING`, `CV_FOLDS`, `INTERACTIVE`, ...).

## 3. Methodology
* **Preprocessing:** drop rows without review or rating (1 to 10); decode HTML entities; remove URLs and wrapping quotes;
  remove duplicates and reviews under 5 words. Stop words are kept because negations carry sentiment.
* **Labels:** rating 1-4 negative, 5-6 neutral, 7-10 positive (a noisy proxy). Classes are imbalanced, so macro-F1 is the main metric.
* **Features:** words after a negator get a `NOT_` prefix; word TF-IDF (1-2 grams) on the negation-marked text plus character
  TF-IDF (2-5 chars) for misspellings. The pipeline takes raw text, so prediction needs no separate preprocessing.
* **Side effects:** a lexicon of 33 categories; a mention is negated if a negator is within the 3 preceding words of the same clause.
* **Models:** Linear SVM, Logistic Regression, Complement Naive Bayes; C or alpha tuned by grid search (macro-F1) on a subsample.
* **Evaluation:** one stratified 80/20 split; models compared by cross-validated macro-F1 on the training set only (the test set
  plays no part in model choice); the selected model is scored once on the test set (accuracy, balanced accuracy, macro
  precision/recall/F1 with a bootstrap 95% interval, macro ROC-AUC, MCC, confusion matrix, ROC curves). A robustness check tests on
  drugs never seen in training.

## 4. Results
Generated from the actual run in the final section of the cell output and saved to `task4_outputs/results.md`.

## 5. Limitations and ethics
Ratings are a rough stand-in for sentiment; reviews are self-selected; the lexicon finds only listed terms; the negation window is
simple; a mentioned symptom does not prove the drug caused it. Do not use for medical decisions.

## 6. Dataset and citation
Grasser, F., Kallumadi, S., Malberg, H., Zaunseder, S. (2018). Aspect-based sentiment analysis of drug reviews applying
cross-domain and cross-data learning. ICDH 2018. https://archive.ics.uci.edu/dataset/462/drug+review+dataset+drugs+com
"""
display(Markdown(README)); (OUT_DIR / "README.md").write_text(README)

# ------------------------------- STEP 1: LOAD DATA -------------------------------
def make_demo_data(n=3000, seed=0):
    """Tiny synthetic stand-in, ONLY used if the real data cannot be loaded."""
    import random; rng = random.Random(seed)
    drugs = ["Sertraline", "Metformin", "Accutane", "Lisinopril", "Zolpidem", "Ibuprofen", "Vyvanse", "Zoloft"]
    se = ["nausea", "headaches", "dizziness", "weight gain", "insomnia", "dry mouth", "fatigue", "rash"]
    pos = ["This medication worked great for me, I feel so much better.", "Life changing, highly recommend it!",
           "Very happy with the results and it works fast."]
    neg = ["This drug was terrible and did not help at all.", "Worst experience ever, I stopped taking it.",
           "Not helpful, I do not recommend it."]
    neu = ["It helped a little but nothing dramatic.", "Mixed feelings, some days are good and some are bad.",
           "Works okay, not great, not terrible."]
    rows = []
    for _ in range(n):
        k = rng.random()
        t, r = (rng.choice(pos), rng.randint(7, 10)) if k < .5 else (rng.choice(neg), rng.randint(1, 4)) if k < .8 else (rng.choice(neu), rng.randint(5, 6))
        for _ in range(rng.randint(0, 2)):
            t += rng.choice([f" I had {rng.choice(se)} for weeks.", f" No {rng.choice(se)} at all."])
        rows.append({"drugName": rng.choice(drugs), "condition": "Unknown", "review": '"' + t + '"', "rating": float(r)})
    return pd.DataFrame(rows)

def read_any(src, name=""):
    sep = "\t" if str(name or src).lower().endswith(".tsv") else None
    buf = io.BytesIO(src) if isinstance(src, (bytes, bytearray)) else src
    return pd.read_csv(buf, sep=sep, engine="python" if sep is None else "c")

def load_drug_reviews():
    if OWN_FILE:
        paths = [OWN_FILE] if isinstance(OWN_FILE, str) else list(OWN_FILE)
        return pd.concat([read_any(p) for p in paths], ignore_index=True)
    for u in UCI_URLS:
        try:
            r = requests.get(u, timeout=120); r.raise_for_status()
            z = zipfile.ZipFile(io.BytesIO(r.content))
            parts = [read_any(z.read(n), n) for n in z.namelist() if n.lower().endswith((".tsv", ".csv"))]
            if parts: print("Downloaded from:", u); return pd.concat(parts, ignore_index=True)
        except Exception as ex:
            print(f"  download failed ({type(ex).__name__}), trying next option...")
    from google.colab import files          # raises ImportError outside Colab
    print("Auto-download failed. Select drugsComTrain_raw.tsv:")
    up = files.upload()
    if not up: raise RuntimeError("No file uploaded")
    return pd.concat([read_any(b, n) for n, b in up.items()], ignore_index=True)

USING_DEMO = False
try:
    raw = load_drug_reviews()
except Exception as ex:
    print(f"Could not load real data ({type(ex).__name__}).\n>>> USING TINY SYNTHETIC DATA: results are NOT real. <<<")
    raw, USING_DEMO = make_demo_data(), True
std = {"drugname": "drugName", "condition": "condition", "review": "review", "rating": "rating"}
raw = raw.rename(columns={c: std[c.lower().strip()] for c in raw.columns if c.lower().strip() in std})
assert {"review", "rating"} <= set(raw.columns), f"Need review + rating columns, found {list(raw.columns)}"
for c in ("drugName", "condition"):
    if c not in raw: raw[c] = "Unknown"
raw = raw[["drugName", "condition", "review", "rating"]]
if len(raw) > MAX_ROWS: raw = raw.sample(MAX_ROWS, random_state=RANDOM_STATE).reset_index(drop=True)
print("\n[1] Loaded:", raw.shape, "| synthetic:", USING_DEMO)

# ------------------------------- STEP 2: RAW-DATA EXPLORATION -------------------------------
print("\n[2] Missing values (%):", raw.isna().mean().mul(100).round(2).to_dict())
print("    Duplicate reviews:", int(raw.duplicated(subset="review").sum()))

# ------------------------------- STEP 3: PREPROCESSING -------------------------------
NEGATORS = {"no", "not", "never", "without", "cannot", "can't", "cant", "won't", "wont", "don't", "dont", "didn't",
            "didnt", "doesn't", "doesnt", "isn't", "isnt", "wasn't", "wasnt", "haven't", "havent", "hasn't", "hasnt",
            "couldn't", "couldnt", "wouldn't", "wouldnt", "nothing", "none", "neither", "nor"}

def clean_review(text):
    s = html.unescape(str(text)).replace("\u2019", "'").replace("\r", " ").replace("\n", " ")
    s = re.sub(r"http\S+|www\.\S+", " ", s).strip().strip('"').strip("'")
    return re.sub(r"\s+", " ", s).strip()

def mark_negation(text, window=3):
    out, left = [], 0
    for tok in re.findall(r"[a-z']+|[.,;:!?]", str(text).lower()):
        if tok in ".,;:!?": left = 0; continue
        t = tok.replace("'", "")
        if tok in NEGATORS: out.append(t); left = window
        elif left > 0: out.append("NOT_" + t); left -= 1
        else: out.append(t)
    return " ".join(out)

df = raw.copy(); n0 = len(df)
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
df = df.dropna(subset=["review", "rating"]); df = df[df["rating"].between(1, 10)]
df["review"] = df["review"].apply(clean_review)
df = df.drop_duplicates(subset="review"); df = df[df["review"].str.split().str.len() >= 5]
df["drugName"] = df["drugName"].fillna("Unknown").astype(str).str.strip()
df["condition"] = df["condition"].fillna("Unknown").astype(str).str.strip()
df.loc[df["condition"].str.contains("</span>|users found this comment", case=False, regex=True), "condition"] = "Unknown"
df["y"] = pd.cut(df["rating"], bins=[0, 4, 6, 10], labels=[0, 1, 2]).astype(int)      # 1-4 neg, 5-6 neu, 7-10 pos
df["sentiment"] = df["y"].map(dict(enumerate(LABELS))); df = df.reset_index(drop=True)
print(f"\n[3] Cleaning: {n0:,} -> {len(df):,} rows | class share:",
      df["sentiment"].value_counts(normalize=True).reindex(LABELS).round(3).to_dict())
print("    negation example:", mark_negation("It did not help and I have no energy."));

# ------------------------------- STEP 4: EDA PLOTS -------------------------------
fig, ax = plt.subplots(1, 3, figsize=(16, 3.8))
df["rating"].value_counts().sort_index().plot(kind="bar", ax=ax[0], color="steelblue"); ax[0].set_title("Rating distribution")
df["sentiment"].value_counts().reindex(LABELS).plot(kind="bar", ax=ax[1], color=["tomato", "goldenrod", "seagreen"])
ax[1].set_title("Sentiment classes (imbalanced)"); ax[1].tick_params(axis="x", rotation=0)
sns.histplot(df["review"].str.split().str.len().clip(upper=300), bins=40, ax=ax[2]); ax[2].set_title("Review length (words)")
plt.tight_layout(); plt.show()

# ------------------------------- STEP 5: SPLIT -------------------------------
X, y = df["review"], df["y"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=TEST_SIZE, stratify=y, random_state=RANDOM_STATE)
Xs, ys = (train_test_split(X_tr, y_tr, train_size=TUNE_SAMPLE, stratify=y_tr, random_state=RANDOM_STATE)[::2]
          if len(X_tr) > TUNE_SAMPLE else (X_tr, y_tr))
print(f"\n[5] Train {len(X_tr):,} | Test {len(X_te):,} (stratified split, seed {RANDOM_STATE})")

# ------------------------------- STEP 6: SIDE-EFFECT LEXICON + FEATURES -------------------------------
LEXICON = {
    "nausea": ["nausea", "nauseous", "nauseated", "queasy"], "vomiting": ["vomiting", "vomit", "threw up", "throwing up"],
    "headache": ["headache", "headaches", "migraine", "migraines"],
    "dizziness": ["dizzy", "dizziness", "lightheaded", "light headed", "vertigo"],
    "fatigue/drowsiness": ["fatigue", "tired", "exhausted", "exhaustion", "lethargic", "drowsy", "drowsiness", "sleepy"],
    "insomnia": ["insomnia", "can't sleep", "cannot sleep", "can not sleep", "could not sleep", "couldn't sleep", "unable to sleep", "trouble sleeping", "sleeplessness"],
    "weight gain": ["weight gain", "gained weight", "gain weight", "gaining weight", "put on weight"],
    "weight loss": ["weight loss", "lost weight", "losing weight"],
    "increased appetite": ["increased appetite", "always hungry", "constant hunger"],
    "loss of appetite": ["loss of appetite", "no appetite", "lost my appetite", "decreased appetite"],
    "anxiety": ["anxiety", "anxious", "panic attack", "panic attacks"],
    "mood swings": ["mood swings", "mood swing", "irritable", "irritability", "angry outbursts"],
    "low mood": ["depressed", "suicidal thoughts", "crying spells"],
    "acne/breakouts": ["acne", "breakouts", "breakout", "pimples"],
    "rash/itching": ["rash", "rashes", "hives", "itching", "itchy"],
    "diarrhea": ["diarrhea", "diarrhoea", "loose stools"], "constipation": ["constipation", "constipated"],
    "stomach pain/cramps": ["stomach pain", "stomach ache", "stomach cramps", "cramps", "cramping", "abdominal pain"],
    "bloating/gas": ["bloating", "bloated", "gas"], "dry mouth": ["dry mouth", "cotton mouth"],
    "hair loss": ["hair loss", "hair falling out", "losing hair", "thinning hair"],
    "sweating": ["sweating", "night sweats", "sweats"], "tremor/shaking": ["tremor", "tremors", "shaking", "jittery", "jitters"],
    "blurred vision": ["blurred vision", "blurry vision"],
    "heart palpitations": ["palpitations", "heart racing", "racing heart", "rapid heartbeat"],
    "muscle/joint pain": ["muscle pain", "muscle aches", "joint pain", "body aches", "achy"],
    "sexual problems": ["low libido", "no sex drive", "loss of libido", "erectile", "sexual side effects"],
    "brain fog/memory": ["brain fog", "memory loss", "memory problems", "foggy", "confusion"],
    "numbness/tingling": ["numbness", "tingling", "numb"], "swelling": ["swelling", "swollen", "edema"],
    "irregular periods": ["spotting", "irregular periods", "breakthrough bleeding"],
    "dry skin": ["dry skin", "dry lips", "chapped lips"], "cough": ["cough", "coughing"],
}
PHRASE2CANON = {p: c for c, ps in LEXICON.items() for p in ps}
SE_PATTERN = re.compile(r"\b(?:" + "|".join(re.escape(p) for p in sorted(PHRASE2CANON, key=len, reverse=True)) + r")\b")
CLAUSE_SPLIT = re.compile(r"[.;!?]|\bbut\b|\bhowever\b|\balthough\b")

def extract_side_effects(text):
    """Return (reported, negated). 'no nausea' -> negated, 'I had nausea' -> reported."""
    reported, negated = set(), set()
    for clause in CLAUSE_SPLIT.split(str(text).lower()):
        for m in SE_PATTERN.finditer(clause):
            before = re.findall(r"[a-z']+", clause[:m.start()].split(",")[-1])[-3:]
            (negated if any(w in NEGATORS for w in before) else reported).add(PHRASE2CANON[m.group(0)])
    return sorted(reported), sorted(negated)

class ReviewPreprocessor(BaseEstimator, TransformerMixin):
    """Stateless: raw strings -> DataFrame(review, review_neg). Lets the saved model accept raw text."""
    def fit(self, X, y=None): return self
    def transform(self, X):
        clean = pd.Series(list(X), dtype=object).fillna("").astype(str).map(clean_review)
        return pd.DataFrame({"review": clean, "review_neg": clean.map(mark_negation)})

MODEL_SPECS = {
    "linear_svm": (lambda: LinearSVC(class_weight="balanced", C=0.1), {"clf__C": [0.05, 0.1, 0.5, 1]}),
    "logistic_regression": (lambda: LogisticRegression(max_iter=2000, class_weight="balanced"), {"clf__C": [0.5, 1, 4]}),
    "complement_nb": (lambda: ComplementNB(), {"clf__alpha": [0.1, 0.3, 1.0]}),
}
def build_pipeline(name):
    feats = ColumnTransformer([
        ("word", TfidfVectorizer(ngram_range=(1, 2), min_df=3, max_features=50000, sublinear_tf=True), "review_neg"),
        ("char", TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=5, max_features=60000, sublinear_tf=True), "review")])
    return Pipeline([("prep", ReviewPreprocessor()), ("features", feats), ("clf", MODEL_SPECS[name][0]())])
print("\n[6] Side-effect demo:", extract_side_effects("No nausea at all, but I got headaches."))

# ------------------------------- STEP 7: TRAINING (tuning + cross-validation) -------------------------------
def get_scores(p, X): return p.predict_proba(X) if hasattr(p, "predict_proba") else p.decision_function(X)
def macro_auc(yt, S):
    Yb = label_binarize(yt, classes=[0, 1, 2]); return float(np.mean([roc_auc_score(Yb[:, k], S[:, k]) for k in range(3)]))
def metrics_dict(yt, pr, S):
    return {"accuracy": accuracy_score(yt, pr), "balanced_accuracy": balanced_accuracy_score(yt, pr),
            "macro_precision": precision_score(yt, pr, average="macro", zero_division=0),
            "macro_recall": recall_score(yt, pr, average="macro", zero_division=0),
            "macro_f1": f1_score(yt, pr, average="macro"), "macro_roc_auc": macro_auc(yt, S), "mcc": matthews_corrcoef(yt, pr)}

skf = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)
summary, fitted, preds, scores = {}, {}, {}, {}
print("\n[7] Training (tuning + cross-validation)...")
for name, (_, grid) in MODEL_SPECS.items():
    pipe, best_params = build_pipeline(name), {}
    if DO_TUNING:
        gs = GridSearchCV(pipe, grid, scoring="f1_macro", cv=3, n_jobs=-1).fit(Xs, ys)
        best_params = {k: (v.item() if hasattr(v, "item") else v) for k, v in gs.best_params_.items()}
        pipe = clone(pipe).set_params(**best_params)
    cv = cross_val_score(clone(pipe), X_tr, y_tr, cv=skf, scoring="f1_macro", n_jobs=-1)
    t0 = time.time(); pipe.fit(X_tr, y_tr); fit_s = time.time() - t0
    pr, S = pipe.predict(X_te), get_scores(pipe, X_te)
    summary[name] = {"best_params": best_params, "cv_f1": float(cv.mean()), "cv_std": float(cv.std()),
                     "test": metrics_dict(y_te, pr, S), "fit_s": round(fit_s, 1)}
    fitted[name], preds[name], scores[name] = pipe, pr, S
    print(f"   {name:20s} params {best_params} | CV macro-F1 {cv.mean():.4f} (+/- {cv.std():.4f}) | fit {fit_s:.0f}s")
best = max(summary, key=lambda n: summary[n]["cv_f1"])
print(f"   Selected model (best CV macro-F1, test set not used): {best}")

# ------------------------------- STEP 8: EVALUATION -------------------------------
best_pipe, best_pred, best_S = fitted[best], preds[best], scores[best]
comp = pd.DataFrame({n: {"cv_macro_f1": v["cv_f1"], **v["test"]} for n, v in summary.items()}).T
print("\n[8] Model comparison:"); display(comp.sort_values("cv_macro_f1", ascending=False).round(4))
print(f"Classification report ({best}, held-out test set):")
print(classification_report(y_te, best_pred, target_names=LABELS, digits=3, zero_division=0))
rng = np.random.RandomState(RANDOM_STATE); yt_a, pr_a = np.asarray(y_te), np.asarray(best_pred); bs = []
for _ in range(N_BOOT):
    i = rng.randint(0, len(yt_a), len(yt_a)); bs.append(f1_score(yt_a[i], pr_a[i], average="macro"))
best_ci = [float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))]
print(f"Macro-F1 95% bootstrap interval: {best_ci[0]:.4f} to {best_ci[1]:.4f}")

fig, ax = plt.subplots(1, 3, figsize=(17, 4.2))
sns.heatmap(confusion_matrix(y_te, best_pred, normalize="true"), annot=True, fmt=".2f", cmap="Blues", vmin=0, vmax=1,
            xticklabels=LABELS, yticklabels=LABELS, ax=ax[0], cbar=False)
ax[0].set_title(f"Confusion matrix: {best.replace('_', ' ')}"); ax[0].set_xlabel("Predicted"); ax[0].set_ylabel("True")
names = list(summary); xs = np.arange(len(names)); w = 0.38
ax[1].bar(xs - w/2, [summary[n]["cv_f1"] for n in names], w, yerr=[summary[n]["cv_std"] for n in names], capsize=3, label="cross-validation")
ax[1].bar(xs + w/2, [summary[n]["test"]["macro_f1"] for n in names], w, label="held-out test")
ax[1].set_xticks(xs); ax[1].set_xticklabels([n.replace("_", " ") for n in names], rotation=15)
ax[1].set_ylim(0, 1); ax[1].set_title("Model comparison (macro-F1)"); ax[1].legend(fontsize=8)
Yb = label_binarize(y_te, classes=[0, 1, 2])
for k, l in enumerate(LABELS):
    fpr, tpr, _ = roc_curve(Yb[:, k], best_S[:, k]); ax[2].plot(fpr, tpr, label=f"{l} (AUC {roc_auc_score(Yb[:, k], best_S[:, k]):.2f})")
ax[2].plot([0, 1], [0, 1], "k--", lw=.8); ax[2].set_title("ROC curves (one-vs-rest)"); ax[2].legend(fontsize=8)
plt.tight_layout(); plt.savefig(OUT_DIR / "evaluation.png", dpi=150); plt.show()

# Robustness: drugs never seen in training
g_tr, g_te = next(GroupShuffleSplit(n_splits=1, test_size=TEST_SIZE, random_state=RANDOM_STATE).split(X, y, groups=df["drugName"]))
unseen_f1 = float(f1_score(y.iloc[g_te], clone(best_pipe).fit(X.iloc[g_tr], y.iloc[g_tr]).predict(X.iloc[g_te]), average="macro"))
print(f"Robustness, macro-F1 on unseen drugs: {unseen_f1:.4f} (random split: {summary[best]['test']['macro_f1']:.4f})")

# Interpretation: strongest features per class
clf = best_pipe.named_steps["clf"]
if hasattr(clf, "coef_"):
    fn = best_pipe.named_steps["features"].get_feature_names_out()
    for k, l in enumerate(LABELS):
        print(f"Top features for {l.upper():9s}:", ", ".join(fn[i].split("__", 1)[-1] for i in np.argsort(-clf.coef_[k])[:10]))

# ------------------------------- STEP 9: SAVE + RELOAD -------------------------------
MODEL_FILE = OUT_DIR / "best_model.joblib"
joblib.dump({"pipeline": best_pipe, "labels": LABELS, "model_name": best, "demo": USING_DEMO}, MODEL_FILE)
bundle = joblib.load(MODEL_FILE)
same = bool((bundle["pipeline"].predict(X_te[:200]) == best_pipe.predict(X_te[:200])).all())
print(f"\n[9] Saved {MODEL_FILE} ({MODEL_FILE.stat().st_size/1e6:.1f} MB); reloaded model predicts identically: {same}")

# ------------------------------- STEP 10: PREDICTION -------------------------------
def confidences(pipe, texts):
    if hasattr(pipe, "predict_proba"): return pipe.predict_proba(texts)
    s = pipe.decision_function(texts); e = np.exp(s - s.max(axis=1, keepdims=True)); return e / e.sum(axis=1, keepdims=True)  # SVM: relative score

def predict_texts(texts, bundle=bundle):
    pipe, labels = bundle["pipeline"], bundle["labels"]; texts = [str(t) for t in texts]
    idx, conf = pipe.predict(texts), confidences(pipe, texts); out = []
    for t, i, c in zip(texts, idx, conf):
        rep, neg = extract_side_effects(t)
        out.append({"review": t, "sentiment": labels[int(i)], "confidence": float(c[int(i)]),
                    "scores": {l: float(v) for l, v in zip(labels, c)}, "reported": rep, "negated": neg})
    return out

def show(r):
    print(f"\n  Review    : {r['review'][:160]}")
    print(f"  Sentiment : {r['sentiment'].upper()} (confidence {r['confidence']:.2f})")
    print("  Scores    : " + ", ".join(f"{k} {v:.2f}" for k, v in r["scores"].items()))
    print("  Side effects reported : " + (", ".join(r["reported"]) or "none"))
    print("  Mentioned but negated : " + (", ".join(r["negated"]) or "none"))

print("\n[10] Example predictions" + (" (SYNTHETIC model, not meaningful)" if USING_DEMO else ""))
for r in predict_texts(["It helped my mood but I get constant nausea and no weight gain.",
                        "Terrible drug, did not work and I could not sleep at all. Stopped after a week.",
                        "Works great for my acne, only a little dry skin at the start."]): show(r)

# ------------------------------- STEP 11: SIDE-EFFECT INSIGHTS -------------------------------
df["side_effects"] = df["review"].map(extract_side_effects).str[0]
n = len(df); overall = df["rating"].mean()
exp = df[["rating", "side_effects"]].explode("side_effects").dropna(subset=["side_effects"])
summ = exp.groupby("side_effects")["rating"].agg(reviews="size", mean_rating="mean")
summ["pct_of_reviews"] = 100 * summ["reviews"] / n; summ["rating_vs_overall"] = summ["mean_rating"] - overall
summ = summ.sort_values("reviews", ascending=False); summ.round(3).to_csv(OUT_DIR / "side_effect_summary.csv")
insights = f"Of {n:,} cleaned reviews, {(df['side_effects'].str.len() > 0).mean():.1%} report at least one side effect."
if len(summ):
    insights += f" The most frequent is {summ.index[0]} ({summ.iloc[0]['pct_of_reviews']:.1f}% of reviews)."
    top = summ.head(10); fig, ax = plt.subplots(1, 2, figsize=(13, 4))
    top["pct_of_reviews"].sort_values().plot.barh(ax=ax[0], color="indianred"); ax[0].set_title("Most reported side effects (% of reviews)")
    top["rating_vs_overall"].sort_values().plot.barh(ax=ax[1], color="steelblue"); ax[1].axvline(0, color="k", lw=.8)
    ax[1].set_title(f"Mean rating when mentioned vs overall ({overall:.2f})")
    plt.tight_layout(); plt.savefig(OUT_DIR / "side_effects.png", dpi=150); plt.show()
    worst = summ[summ["reviews"] >= max(20, 0.002 * n)].sort_values("rating_vs_overall").head(1)
    if len(worst):
        insights += (f" Among effects with enough mentions, {worst.index[0]} has the lowest mean rating "
                     f"({worst.iloc[0]['mean_rating']:.2f}, {worst.iloc[0]['rating_vs_overall']:+.2f} vs overall). "
                     "This is an association in self-reported text, not proof of causation.")

# ------------------------------- STEP 12: RESULTS SUMMARY (auto-written from the run) -------------------------------
bt = summary[best]["test"]; rep = classification_report(y_te, best_pred, target_names=LABELS, output_dict=True, zero_division=0)
weakest = min(LABELS, key=lambda l: rep[l]["f1-score"]); share = {l: float((df["y"] == i).mean()) for i, l in enumerate(LABELS)}
L = []
if USING_DEMO: L.append("> **WARNING: SYNTHETIC demo data; these numbers mean nothing. Run on the real UCI data.**\n")
L.append(f"**Data.** {len(df):,} cleaned reviews (from {len(raw):,} raw rows), {len(X_tr):,} train / {len(X_te):,} test. "
         f"Class shares: {share['negative']:.1%} negative, {share['neutral']:.1%} neutral, {share['positive']:.1%} positive, so macro-F1 is the main metric.\n")
L.append("**Model comparison.** " + "; ".join(f"{k.replace('_', ' ')}: CV macro-F1 {v['cv_f1']:.3f} (sd {v['cv_std']:.3f}), test macro-F1 {v['test']['macro_f1']:.3f}" for k, v in summary.items()) + ".\n")
L.append(f"**Selected model.** {best.replace('_', ' ')} (highest cross-validated macro-F1). Test set: accuracy {bt['accuracy']:.3f}, balanced accuracy "
         f"{bt['balanced_accuracy']:.3f}, macro-F1 {bt['macro_f1']:.3f} (95% interval {best_ci[0]:.3f} to {best_ci[1]:.3f}), macro ROC-AUC {bt['macro_roc_auc']:.3f}, MCC {bt['mcc']:.3f}.\n")
L.append("**Per-class.** " + "; ".join(f"{l} F1 {rep[l]['f1-score']:.3f} (P {rep[l]['precision']:.3f}, R {rep[l]['recall']:.3f})" for l in LABELS)
         + f". Weakest class: {weakest}." + (" Expected: neutral wording overlaps with both neighbours." if weakest == "neutral" else "") + "\n")
L.append(f"**Robustness.** On drugs unseen in training, macro-F1 was {unseen_f1:.3f} vs {bt['macro_f1']:.3f} on the random split.\n")
L.append(f"**Side-effect insights.** {insights}\n")
L.append("Figures saved to `task4_outputs/evaluation.png` and `task4_outputs/side_effects.png`.")
results_md = "\n".join(L); display(Markdown("## Results summary\n" + results_md))
(OUT_DIR / "results.md").write_text(results_md)
(OUT_DIR / "metrics.json").write_text(json.dumps({"best_model": best, "best_ci": best_ci, "unseen_drug_macro_f1": unseen_f1,
                                                  "models": summary, "demo": USING_DEMO}, indent=2, default=float))
print("Saved:", sorted(os.listdir(OUT_DIR)))
try:
    import shutil; shutil.make_archive("task4_outputs", "zip", OUT_DIR)
    from google.colab import files; files.download("task4_outputs.zip")
except Exception:
    pass

# ------------------------------- STEP 13: USER INPUT -------------------------------
if INTERACTIVE:
    print("\n" + "=" * 62 + "\nTry it yourself. Type a drug review and press Enter ('q' or empty to stop).\n"
          "(Research/education only, not medical advice.)\n" + "=" * 62)
    while True:
        try: text = input("Your review > ").strip()
        except EOFError: break
        if text.lower() in ("", "q", "quit", "exit"): print("Done."); break
        if len(text.split()) < 3: print("Please write at least a few words."); continue
        show(predict_texts([text])[0])
```
